#### Experimento de raios catódicos para determinação de $e/m_{e}$

```python
import numpy as np
from numpy.polynomial import Polynomial
import matplotlib.pyplot as plt
from matplotlib import rcParams
```

```python
data = np.load("raios_catodicos_data.npz", allow_pickle=True)
```

```python
rcParams.update({
    "text.usetex": True,
    "font.family": "serif", 
    "font.serif": ["Computer Modern Roman"],  # default LaTeX font
    "axes.labelsize": 16,    # LaTeX font size
    "font.size": 15,
    "legend.fontsize": 15,
    "xtick.labelsize": 16,
    "ytick.labelsize": 16
})
```

```python
x1_plot = data['x1_plot']
y1_plot = data['y1_plot']

x2_plot = data['x2_plot']
y2_plot = data['y2_plot']

x3_plot = data['x3_plot']
y3_plot = data['y3_plot']

x4_plot = data['x4_plot']
y4_plot = data['y4_plot']

x_r1 = data['x_r1']
y_r1 = data['y_r1'] 

x_r2 = data['x_r2']
y_r2 = data['y_r2']

x_r3 = data['x_r3']
y_r3 = data['y_r3']

x_r4 = data['x_r4']
y_r4 = data['y_r4']

a1 = data['a1']
a2 = data['a2']
a3 = data['a3']
a4 = data['a4']


unc_I = 0.01
unc_U = 1

#parâmetros numéricos
mu0 = 4*np.pi*10**-7
R_b = 0.20 #(m)
N_voltas = 154
r1 = 0.02 #(m)
r2 = 0.03
r3 = 0.04
r4 = 0.05


k = lambda r: (r**2 / 2) * 64/125 * (mu0*N_voltas/R_b)**2
```

```python
plt.figure(figsize=(14,10))

plt.subplot(2,2,1)
plt.plot(x1_plot, y1_plot, label= r"$e/m_e =$"+ f" {a1*10**-11:.2f}" + r" $\times$ " + r"$10^{11}$" + r" $\frac{C}{Kg}$", color="red", linestyle="dashed")
plt.errorbar(x_r1, y_r1, yerr=unc_U, xerr=k(r1)*unc_I, color="black", capsize=6, fmt="o")
plt.ylabel(r"$U$ (V)")
#plt.xlabel(r"$kI^{2}$ (C V/Kg)")
plt.title(r"$r = 0.02 m$")
plt.tight_layout()
plt.legend()

plt.subplot(2,2,2)
plt.plot(x2_plot, y2_plot, label= r"$e/m_e =$"+ f" {a2*10**-11:.2f}" + r" $\times$ " + r"$10^{11}$" + r" $\frac{C}{Kg}$", color="red", linestyle="dashed")
plt.errorbar(x_r2, y_r2, yerr=unc_U, xerr=k(r2)*unc_I, color="black", capsize=6, fmt="o")
plt.ylabel(r"$U$ (V)")
plt.xlabel(r"$kI^{2}$ (C V/Kg)")
plt.title(r"$r = 0.03 m$")
plt.tight_layout()
plt.legend()

plt.subplot(2,2,3)
plt.plot(x3_plot, y3_plot, label= r"$e/m_e =$"+ f" {a3*10**-11:.2f}" + r" $\times$ " + r"$10^{11}$" + r" $\frac{C}{Kg}$", color="red", linestyle="dashed")
plt.errorbar(x_r3, y_r3, yerr=unc_U, xerr=k(r3)*unc_I, color="black", capsize=6, fmt="o")
plt.ylabel(r"$U$ (V)")
plt.xlabel(r"$kI^{2}$ (C V/Kg)")
plt.title(r"$r = 0.04 m$")
plt.tight_layout()
plt.legend()

plt.subplot(2,2,4)
plt.plot(x4_plot, y4_plot, label= r"$e/m_e =$"+ f" {a4*10**-11:.2f}" + r" $\times$ " + r"$10^{11}$" + r" $\frac{C}{Kg}$", color="red", linestyle="dashed")
plt.errorbar(x_r4, y_r4, yerr=unc_U, xerr=k(r4)*unc_I, color="black", capsize=6, fmt="o")
plt.ylabel(r"$U$ (V)")
plt.xlabel(r"$kI^{2}$ (C V/Kg)")
plt.title(r"$r = 0.05 m$")
plt.tight_layout()
plt.legend()

plt.show()
```
